# Green-Kubo methods

The Green-Kubo (GK) methods are a family of equilibrium simulation analyses that can be used to estimate different transport properties (diffusion, viscosity, and thermal conductivity) of a material. At the center of all of the GK methods is an integral of an auto-correlation function computed from a measured property. Evaluating the integral is the trickest part of the GK methods.

## Auto-correlation functions

The auto-correlation function ($acf$) of some property $X(t)$ is $\left<X_0 \cdot X_t\right>$, where $X_0$ and $X_t$ are measurements of $X(t)$ taken at different times separated by some $t$.  $acf(X(t))$ can then be evaluated by varying the $t$ from which $\left<X_0 \cdot X_t\right>$ is evaluated. As a note of caution, $acf(X(t))$ can be evaluated differently across different fields, so it is important that the evaluation is done in the correct way for GK calculations!  Notably, $acf(t)$ should not be normalized and have units that are consistent with $X^2$, and measurements should be divided by the number of samples, i.e. be sure it is $\left<X_0 \cdot X_t\right>$ and not just $X_0 \cdot X_t$.

The $acf(X(t))$ function has some interesting properties that need to be considered for the GK calculations.  First, it is expected that there is some value of $t$ beyond which $acf(X(t)) = 0$ resulting in the integral of $acf(X(t))$ being finite. Ideally, the integral of the computed $acf(X(t))$ should converge to a constant with increasing maximum $t$. However, it is more typical to see the computed $acf(X(t))$ at large $t$ fluctuating around 0 or near 0, which introduces error in the integral calculation and can cause it to deviate with increasing maximum $t$.

Second, when evaluating time series data where $X(t)$ is sampled at regular intervals, each new sample provides an additional $X_0 \cdot X_t$ measurement for all $t$ intervals less than or equal to the current total run time.
